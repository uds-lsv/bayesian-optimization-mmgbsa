```python
import pandas as pd
import numpy as np
import seaborn as sns

experimental = pd.read_csv("../experimental_results.csv")
md = pd.read_csv("../benchmark.csv")
md["name"] = md["name"].str.removeprefix("ligand_benchmark_")
md["name"] = md["name"].map(BENCHMARK_LOOKUP)
md = md.rename(columns={"name": "smiles"})

md_agg = md.groupby("smiles")["dg_en_gb"].agg(["mean", "sem"]).rename(columns={"mean": "dg_en_gb", "std": "se"})
```

```python
# conversion requires openff-units
# conda install -c conda-forge openff-units
from openff.units import unit

def convert_ki_to_dg(value: float):
    q = unit.Quantity(value, unit.micromolar)
    temp = unit.Quantity(300, unit.kelvin)
    energy = (unit.molar_gas_constant* temp * np.log(q / unit.molar)).to("kcal/mol")
    return energy.magnitude

def convert_error_to_dg(value, error):
    """
    :param prod: Product of value and error of the measurement
    """
    q = unit.Quantity(value, unit.micromolar)
    qerr = unit.Quantity(error, unit.micromolar)
    temp = unit.Quantity(300, unit.kelvin)
    error = (unit.molar_gas_constant*300*unit.kelvin * qerr/ q).to("kcal/mol")
    return error.magnitude
```

```python
df = pd.merge(experimental, md_agg, on="smiles")
df["exp_dg"] = df["ki"].apply(convert_ki_to_dg)
df["exp_error_dg"] = df.apply(lambda r: convert_error_to_dg(r["ki"], r["error"]), axis=1)
df.head()
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots()

coef = np.polyfit(df["dg_en_gb"], df["exp_dg"], 1)
regression = np.poly1d(coef)
x = np.arange(-58, -29)
ax.plot(x, regression(x), "k--", label="Regression line")
ax.fill_between(x, regression(x) - 2, regression(x) + 2, alpha=0.2, label="± 2 kcal/mol")
ax.fill_between(x, regression(x) - 1, regression(x) + 1, alpha=0.3, label="± 1 kcal/mol")

ax.errorbar(df["dg_en_gb"], df["exp_dg"], xerr=df["sem"], yerr=df["exp_error_dg"], fmt=".")

ax.set_ylabel("Experimental")
ax.set_xlabel("MMPBSA")
ax.legend()
ax.set_title("Experiment vs MMPBSA Simulation (kcal/mol)")

ax.set_xlim(-58, -30)
fig.set_size_inches(5, 5)
```
